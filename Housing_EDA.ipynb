```python
import matplotlib.pyplot as plt
import pandas as pd

##Finding Patteern (Comparison of river location vs house price)
df["PRICE_LEVEL"] = (df["MEDV"] > df["MEDV"].median()).astype(int)

pd.crosstab(df["CHAS"], df["PRICE_LEVEL"]).plot(kind="bar",
                                                 color=["salmon", "lightblue"])
plt.title("Comparison of River Location to House Price Level")
plt.xlabel("0 = Not near river, 1 = Near river")
plt.ylabel("Total")
plt.legend(["Low price", "High price"])
plt.xticks(rotation=0);
```

```python
import pandas as pd
import matplotlib as plt

df = pd.read_excel("Documents/housing.ass.xlsx")
```

```python
import matplotlib.pyplot as plt
```

```python
##Checking for Correlation between Rooms and House Value
plt.figure(figsize=(10, 6))

plt.scatter(df.RM[df.CHAS==0],
            df.MEDV[df.CHAS ==0],
            color="lightblue")

plt.scatter(df.RM[df.CHAS > 0],
            df.MEDV[df.CHAS > 0],
            color="salmon")

plt.title("House Value in function of Rooms")
plt.xlabel("Average Number of Rooms")
plt.ylabel("Median Home Value")
plt.legend(["Not Near river", "Near river"]);
```

```python
df["CHAS"].value_counts()
```

```python
df["PRICE_LEVEL"] = (df["MEDV"] > df["MEDV"].median()).astype(int)

pd.crosstab(df["RAD"], df["PRICE_LEVEL"]).plot(kind="bar",
                                               figsize=(10, 6),
                                               color=["salmon", "lightblue"])

plt.xlabel("Highway Access (RAD)")
plt.ylabel("Amount")
plt.title("House Price Level per Highway Access")
plt.legend(["Low price", "High price"])
plt.xticks(rotation=45);
```
